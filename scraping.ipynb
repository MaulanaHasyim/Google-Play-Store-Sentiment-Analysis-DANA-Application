```python
import pandas as pd
from google_play_scraper import Sort, reviews
from google.colab import files

app_id = 'id.dana'
all_reviews = []
target_per_star = 2500

# Scraping data ulasan per rating bintang
for star in range(1, 6):
    try:
        result, _ = reviews(
            app_id,
            lang='id',
            country='id',
            sort=Sort.NEWEST,
            count=target_per_star,
            filter_score_with=star
        )
        all_reviews.extend(result)
    except Exception as e:
        print(f"Error pada bintang {star}: {e}")

# Convert ke dataframe dan seleksi kolom
df = pd.DataFrame(all_reviews)
df_selected = df[['userName', 'content', 'score']].copy()

# Pelabelan sentimen berdasarkan score bintang
def label_sentiment(score):
    if score in [1, 2]:
        return 'Negatif'
    elif score == 3:
        return 'Netral'
    else:
        return 'Positif'

df_selected['sentiment'] = df_selected['score'].apply(label_sentiment)

# Menampilkan informasi dataset
print("Distribusi Kelas:")
print(df_selected['sentiment'].value_counts())
print(f"Total Data: {len(df_selected)}")

# Simpan dan download dataset
output_filename = 'dataset_playstore_dana.csv'
df_selected.to_csv(output_filename, index=False, encoding='utf-8')
files.download(output_filename)
```
